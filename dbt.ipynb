```python
import glob

# Find all SQL files in your staging folder
sql_files = glob.glob('models/staging/*.sql')

for file_path in sql_files:
    # Read the file using utf-8-sig (this strips the invisible BOM character)
    with open(file_path, 'r', encoding='utf-8-sig') as f:
        content = f.read()
    
    # Save the file back as standard, clean utf-8
    with open(file_path, 'w', encoding='utf-8') as f:
        f.write(content)
        
print("BOMs successfully destroyed! Files are clean.")
```
